```python
#Primeiro código da aula de FCM2026

import matplotlib.pyplot as plt

# 1. Define os dados (4 pontos: (1, 2), (2, 4), (3, 6), (4, 8))
x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

# 2. Cria o gráfico
plt.plot(x, y, marker='o', color='b', linestyle='--')

# 3. Adiciona títulos e rótulos
plt.title('Gráfico Simples de 4 Pontos')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

# 4. Exibe a grade e o gráfico
plt.grid(True)
plt.show()

#Gráfico 2

# 1. Define os dados das 4 barras
w = ['Ponto A', 'Ponto B', 'Ponto C', 'Ponto D'] # Podem ser números [1, 2, 3, 4] ou categorias
z = [2, 4, 6, 8]

# 2. Cria o gráfico de barras
plt.bar(w, z, color='skyblue', edgecolor='black')

# 3. Adiciona títulos e rótulos
plt.title('Gráfico de Barras com 4 Pontos')
plt.xlabel('Categorias')
plt.ylabel('Valores')

# 4. Exibe o gráfico
plt.show()

#comentário para outra commit
```
